# Search Ranking Model — Data Exploration
Explore queries, documents and relevance labels.

```python
from src.data_loader import load_dataset
import matplotlib.pyplot as plt
import seaborn as sns

df = load_dataset()
df.head()
```

```python
print(df.info())
print('\nRelevance counts:')
print(df['relevance'].value_counts().sort_index())
print('\nDocuments per query:')
print(df.groupby('query_id').size())
```

```python
sns.countplot(data=df, x='relevance')
plt.title('Relevance Distribution')
plt.show()
```
