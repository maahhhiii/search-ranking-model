# TF-IDF Baseline
Use TF-IDF cosine similarity as a simple ranking baseline.

```python
from src.data_loader import load_dataset
from src.preprocessing import preprocess_dataset
from src.features import build_features
from sklearn.model_selection import GroupShuffleSplit
from sklearn.metrics import ndcg_score
import numpy as np

df = preprocess_dataset(load_dataset())
train_idx, test_idx = next(GroupShuffleSplit(n_splits=1, test_size=.25, random_state=42).split(df, groups=df.query_id))
test = df.iloc[test_idx].copy()
X, vectorizer = build_features(test, fit_vectorizer=True)
test['score'] = X['tfidf_similarity']
```

```python
values = []
for _, g in test.groupby('query_id'):
    values.append(ndcg_score([g.relevance.values], [g.score.values], k=5))
print('Baseline mean NDCG@5:', np.mean(values))
```
